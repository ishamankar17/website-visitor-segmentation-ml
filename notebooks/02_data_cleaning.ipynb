# 02 · Data Cleaning
We apply `src.data_preprocessing.clean_data` and save `data/processed/visitors_cleaned.csv`.

```python
from src.data_preprocessing import load_raw, clean_data, save_processed

raw = load_raw()
print('raw     :', raw.shape)
print('dups    :', raw.duplicated().sum())
print('months  :', raw.Month.unique())
```

```python
clean = clean_data(raw)
print('cleaned :', clean.shape)
print('months  :', clean.Month.unique())
print('missing :', clean.isna().sum().sum())
clean.head()
```

```python
assert (clean.select_dtypes('number') >= 0).all().all()
assert clean[['BounceRates','ExitRates']].max().max() <= 1
print('Sanity checks passed')
```

```python
path = save_processed(clean)
print('Saved to', path)
```

**What changed:** 125 duplicate sessions dropped, `June`→`Jun`, numeric `Month_Num` added, `Weekend`/`Revenue` converted to 0/1, plus defensive imputation and clipping (no-ops on this file, but protect against future data).
